## Define positions of the Dobot

To define the position of the Dobot, you can use the following code. With the values x, y, z and r you can define different positions for the Dobot and move it's arm.

**Important notes**:
- To start the code the Dobot must be on and the light on the robot needs to be green.
- To ensure that the Dobot moves reliably to the pre-programmed positions, it is important that it always starts from the same starting position at the beginning of the program. That is why it is important to “home” the Dobot before each run, or at least after adjusting its settings. This is done with the function `init_and_home_dobot`. 
- The position of the Dobot arm must not be adjusted manually. If this does occur, the Dobot must be returned to its home position using the `init_and_home_dobot` function.

### Dobot's coordinate system

| Parameter  | Direction | Unit |
| ------------- | ------------- | ------------- |
| X  | Back and forth  | mm |
| Y  | Left and right  | mm |
| Z  | Height  | mm |
| R  | Radius  | degree |

First you need to establish a connection to the Dobot. Here you may need to change the number of the port. 

```python
import time
from helper_functions.dobot_functions import find_dobot_ports, init_and_home_dobot, safe_move

ports = find_dobot_ports()
print(ports)

# choose the correct port
dobot = init_and_home_dobot(ports[1])

time.sleep(1)

print("dobot:", dobot)
```

Now you can define different positions for the Dobot and store them as variables.

```python
pick_color_sensor_sorter = (150, -190, 50, 65)
throw_position_sorter = (250, 100, 50, 0)
conveyor_position_sorter = (230, -50, 50, 0)
pick_conveyor_belt_pick_place = (230, 85, 30, 55)
place_color_sensor_pick_place = (150, 255, 50, 45)
```

Now you can move the Dobot to a positon, which you defined before.

The **safe_move** function is used, that the Dobot does not crash into something if he moves. With that function the Dobot first moves to a safe height, then moves to the desired x and y coordinates and finally moves down to the desired z coordinate. 

```python
safe_move(dobot, place_color_sensor_pick_place)

time.sleep(1)

dobot.interface.close_col()
```
