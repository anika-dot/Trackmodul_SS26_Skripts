## Test color scanner

To test the color sensor's functionality, you can use the following code. The color scanner must be connected to a Dobot.

The first step is to establish a connection to the Dobot. You may need to change the port to do this. Also, the color sensor is activated. The `print` statement is displayed to verify that the connection was established.

```python
import json
import time
from turtle import color
from xml.etree.ElementTree import VERSION
from xmlrpc import client
from helper_functions.dobot_functions import find_dobot_ports
from pydobotplus import Dobot

ports = find_dobot_ports()
color_sensor = Dobot(port=ports[1])
color_sensor.set_color(True, VERSION)
print("Color sensor ready")
```

Now the scanning operation can begin. To do this, place a block on the color sensor. 

The color is output in RGB code. The color sensor can only determine the colors and return the value “True” or “False” for red, green, and blue. You then receive a list containing the values for the corresponding color.

```python
print("Reading color sensor...")
color_sensor.get_color()
rgb = color_sensor.get_color()
time.sleep(0.3)
print("Color sensor finished scanning")
print(rgb)
```

To check whether the color blue has been detected and to return the corresponding value, you need to implement a code something like this:

```python
if rgb == [False, False, True]:
    color = "blue"
else:
    color = "other"

print(f"Detected color: {color}")
```
